**Combine CSVs from Subfolders into One DataFrame**

```
import pandas as pd
import glob

files = glob.glob('main_folder/**/*.csv', recursive=True)
df = pd.concat((pd.read_csv(f) for f in files), ignore_index=True)
```



*   glob(..., recursive=True) finds CSVs in all nested subfolders
*   pd.concat stacks them into one DataFrame
*   ignore_index=True keeps row index clean



Optional — track source file per row:


`df = pd.concat((pd.read_csv(f).assign(source_file=f) for f in files), ignore_index=True)`
